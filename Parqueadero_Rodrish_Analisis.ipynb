## 📋 Descripción del Problema

La empresa logística **Rodrish S.A.** necesita optimizar el diseño de su parqueadero de maniobras para camiones.

### Datos Conocidos:
- ✅ Rectángulo A: 20m × 10m = 200 m²
- ✅ Rectángulo B: 15m × 10m = 150 m²
- ✅ Rectángulo C: 20m × 10m = 200 m²
- ✅ Radio de conos: 40 cm
- ✅ Distanciamiento: 2 m
- ✅ Largo camión: 12 m
- ✅ Ampliación máxima: 50 m²
- ✅ Camiones requeridos: 15

## 🎯 PARTE 1: DATOS Y CÁLCULOS

```python
# Definir datos del problema
class Datos:
    # Rectángulos
    RECT_A = 20 * 10  # 200 m²
    RECT_B = 15 * 10  # 150 m²
    RECT_C = 20 * 10  # 200 m²
    AREA_TOTAL = RECT_A + RECT_B + RECT_C  # 550 m²
    
    # Camión y seguridad
    CAMION_LARGO = 12
    CAMION_ANCHO = 2.5
    DISTANCIAMIENTO = 2
    RADIO_CONO = 0.4
    
    # Requerimientos
    NUM_CAMIONES = 15
    AREA_ADICIONAL = 50
    
    # Cálculos
    ESPACIO_LARGO = CAMION_LARGO + 2 * DISTANCIAMIENTO  # 16m
    ESPACIO_ANCHO = CAMION_ANCHO + 2 * DISTANCIAMIENTO  # 6.5m
    AREA_POR_CAMION = ESPACIO_LARGO * ESPACIO_ANCHO  # 104 m²
    AREA_NECESARIA = NUM_CAMIONES * AREA_POR_CAMION  # 1560 m²
    DEFICIT = AREA_NECESARIA - AREA_TOTAL  # 1010 m²
    PERIMETRO = 110

datos = Datos()

# Mostrar resumen
print("="*70)
print("           RESUMEN DE DATOS Y CÁLCULOS")
print("="*70)
print(f"\n📐 ÁREA TOTAL ACTUAL: {datos.AREA_TOTAL} m²")
print(f"📏 PERÍMETRO: {datos.PERIMETRO} m")
print(f"\n🚛 ESPACIO POR CAMIÓN: {datos.AREA_POR_CAMION:.1f} m²")
print(f"   • Largo con distanciamiento: {datos.ESPACIO_LARGO} m")
print(f"   • Ancho con distanciamiento: {datos.ESPACIO_ANCHO} m")
print(f"\n📊 ÁREA NECESARIA TOTAL: {datos.AREA_NECESARIA:.1f} m²")
print(f"⚠️  DÉFICIT: {datos.DEFICIT:.1f} m²")
print(f"\n💡 CAPACIDAD ACTUAL: {datos.AREA_TOTAL/datos.AREA_POR_CAMION:.1f} camiones")
print(f"🎯 REQUERIMIENTO: {datos.NUM_CAMIONES} camiones")
print("\n" + "="*70)
```

```python
# ====================================================================================
# FUNCIONES DE DIBUJO Y VALIDACIÓN DE CAMIONES
# ====================================================================================

def validar_camion_en_limites(x, y, angulo, limites):
    """
    Valida que un camión y su zona de seguridad estén completamente dentro de los límites del rectángulo.
    
    Parámetros:
    - x, y: Coordenadas del centro del camión
    - angulo: Ángulo de rotación (0, 90, etc.)
    - limites: dict con keys 'x_min', 'x_max', 'y_min', 'y_max'
    
    Retorna:
    - True si el camión está completamente dentro de los límites, False en caso contrario
    """
    import numpy as np
    
    # Dimensiones del camión con zona de seguridad
    ESPACIO_LARGO = 16  # 12m + 2*2m
    ESPACIO_ANCHO = 6.5  # 2.5m + 2*2m
    
    if angulo == 0:  # Horizontal
        x_min_zona = x - ESPACIO_LARGO/2
        x_max_zona = x + ESPACIO_LARGO/2
        y_min_zona = y - ESPACIO_ANCHO/2
        y_max_zona = y + ESPACIO_ANCHO/2
    elif angulo == 90:  # Vertical
        x_min_zona = x - ESPACIO_ANCHO/2
        x_max_zona = x + ESPACIO_ANCHO/2
        y_min_zona = y - ESPACIO_LARGO/2
        y_max_zona = y + ESPACIO_LARGO/2
    else:
        # Para otros ángulos, usar cálculo general
        half_largo = ESPACIO_LARGO / 2
        half_ancho = ESPACIO_ANCHO / 2
        theta = np.radians(angulo)
        
        # Calcular los 4 vértices rotados
        vertices = [
            (-half_largo, -half_ancho),
            (half_largo, -half_ancho),
            (half_largo, half_ancho),
            (-half_largo, half_ancho)
        ]
        
        vertices_rotados = []
        for vx, vy in vertices:
            rx = vx * np.cos(theta) - vy * np.sin(theta) + x
            ry = vx * np.sin(theta) + vy * np.cos(theta) + y
            vertices_rotados.append((rx, ry))
        
        x_coords = [v[0] for v in vertices_rotados]
        y_coords = [v[1] for v in vertices_rotados]
        
        x_min_zona = min(x_coords)
        x_max_zona = max(x_coords)
        y_min_zona = min(y_coords)
        y_max_zona = max(y_coords)
    
    # Validar que la zona esté dentro de los límites
    dentro = (x_min_zona >= limites['x_min'] and 
              x_max_zona <= limites['x_max'] and
              y_min_zona >= limites['y_min'] and
              y_max_zona <= limites['y_max'])
    
    if not dentro:
        print(f"⚠️ ADVERTENCIA: Camión en ({x}, {y}) con ángulo {angulo}° está FUERA de límites")
        print(f"   Zona: x=[{x_min_zona:.2f}, {x_max_zona:.2f}], y=[{y_min_zona:.2f}, {y_max_zona:.2f}]")
        print(f"   Límites: x=[{limites['x_min']}, {limites['x_max']}], y=[{limites['y_min']}, {limites['y_max']}]")
    
    return dentro


def dibujar_zona_seguridad(ax, x, y, angulo):
    """
    Dibuja la zona de seguridad alrededor del camión (área amarilla de 16m × 6.5m).
    
    Parámetros:
    - ax: Axes de matplotlib
    - x, y: Coordenadas del centro del camión
    - angulo: Ángulo de rotación en grados (0, 90, etc.)
    """
    import numpy as np
    from matplotlib.patches import Polygon
    
    ESPACIO_LARGO = 16  # 12m + 2*2m
    ESPACIO_ANCHO = 6.5  # 2.5m + 2*2m
    
    # Definir los 4 vértices del rectángulo de seguridad (sin rotar)
    half_largo = ESPACIO_LARGO / 2
    half_ancho = ESPACIO_ANCHO / 2
    
    vertices = np.array([
        [-half_largo, -half_ancho],
        [half_largo, -half_ancho],
        [half_largo, half_ancho],
        [-half_largo, half_ancho]
    ])
    
    # Rotar los vértices
    theta = np.radians(angulo)
    rotation_matrix = np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta), np.cos(theta)]
    ])
    
    vertices_rotados = vertices @ rotation_matrix.T
    
    # Trasladar al centro (x, y)
    vertices_rotados[:, 0] += x
    vertices_rotados[:, 1] += y
    
    # Dibujar el polígono
    zona = Polygon(vertices_rotados, facecolor='yellow', edgecolor='orange',
                   alpha=0.3, linestyle='--', linewidth=2)
    ax.add_patch(zona)


def dibujar_camion_exacto(ax, x, y, angulo, color, num_camion, validar=False, limites=None):
    """
    Dibuja un camión con rotación exacta usando polígonos.
    
    Parámetros:
    - ax: Axes de matplotlib
    - x, y: Coordenadas del centro del camión
    - angulo: Ángulo de rotación en grados (0 = horizontal, 90 = vertical)
    - color: Color del camión (hex o nombre)
    - num_camion: Número del camión para la etiqueta
    - validar: Si True, valida que el camión esté dentro de límites
    - limites: dict con 'x_min', 'x_max', 'y_min', 'y_max' (requerido si validar=True)
    """
    import numpy as np
    from matplotlib.patches import Polygon
    
    # Validar si se solicita
    if validar and limites:
        if not validar_camion_en_limites(x, y, angulo, limites):
            # Dibujar en rojo si está fuera de límites
            color = '#ff0000'
    
    CAMION_LARGO = 12
    CAMION_ANCHO = 2.5
    
    # Definir los 4 vértices del camión (sin rotar, centrado en origen)
    half_largo = CAMION_LARGO / 2
    half_ancho = CAMION_ANCHO / 2
    
    vertices = np.array([
        [-half_largo, -half_ancho],
        [half_largo, -half_ancho],
        [half_largo, half_ancho],
        [-half_largo, half_ancho]
    ])
    
    # Aplicar rotación
    theta = np.radians(angulo)
    rotation_matrix = np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta), np.cos(theta)]
    ])
    
    vertices_rotados = vertices @ rotation_matrix.T
    
    # Trasladar al centro (x, y)
    vertices_rotados[:, 0] += x
    vertices_rotados[:, 1] += y
    
    # Dibujar el polígono del camión
    camion = Polygon(vertices_rotados, facecolor=color, edgecolor='black',
                     alpha=0.8, linewidth=2)
    ax.add_patch(camion)
    
    # Agregar etiqueta con el número del camión
    ax.text(x, y, str(num_camion), ha='center', va='center',
            fontsize=12, fontweight='bold', color='white',
            bbox=dict(boxstyle='circle', facecolor='black', alpha=0.7, pad=0.3))

print("✅ Funciones de dibujo y validación creadas correctamente")
```
